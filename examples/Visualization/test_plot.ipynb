```python
import qewton
import pandas as pd
import json
from pathlib import Path
import numpy as np


config_path = "./test_tune_1/setup.json"

data = pd.read_csv("./test_tune_1/results.csv")
with Path(config_path).open("r", encoding="utf-8") as f:
    config_data: dict = json.load(f)
```

```python
table_plot = qewton.viz.TableScatter(data=data, 
                                     axis_keys=config_data["Hyperparameters"], 
                                     objective_keys=config_data["Objective Names"])
fig = qewton.viz.Figure(table_plot)#.show()
fig.show()
```

```python
dash_app = qewton.viz.DashApplication().create(fig)
dash_app.run(debug=True, port=8050)
```
